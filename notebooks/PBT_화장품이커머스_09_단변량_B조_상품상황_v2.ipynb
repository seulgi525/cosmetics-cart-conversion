### 4-1. 판정에 필요한 칸만 추린다

````python
pv = B_SUM[["mean", "50%", "max", "iqr", "lower_bound", "upper_bound",
            "skew", "kurt", "upper_outliers", "lower_outliers", "outliers_ratio"]].copy()

pv["로그변환 후보"] = pv["skew"].abs() >= 1.0          # 우리 기준 : 1.0
pv["outliers_ratio"] = (pv["outliers_ratio"] * 100).round(2)
pv.rename(columns={"outliers_ratio": "이상치%"}, inplace=True)

pv.round(3)
````

> **`lower_bound`가 음수다.** 가격은 음수가 될 수 없으므로 **하한 이상치는 0건**이고,
> 세어진 이상치는 **전량 상한(비싼 상품)**이다. `upper_outliers`와 `lower_outliers`로 확인된다.

### 4-3. 봉우리 수와 이산점(군집) — **판정 근거** 〈추가된 셀 · v2에서 보강〉

점검 항목 ②(봉우리 수)와 ④(이산점)는 요약 통계로는 안 보인다.

- **이산점** : 값의 종류가 몇 개이고, 상위 몇 개 값이 전체의 얼마를 덮는가. 값이 어느 자리까지 떨어지는가
- **봉우리** : 원척도는 오른쪽 꼬리가 너무 길어 형태가 안 보이므로 **log 척도에서도 함께** 본다
- **꼬리** : `numerical_summary`는 25/50/75분위만 주므로 **상위 분위수를 따로** 구한다

> **v2에서 바뀐 점** — v1은 이 셀이 12구간 히스토그램만 찍으면서, 판정 문장에는
> **다른 구간 설정(20구간)으로 계산한 수치**와 **하드코딩된 "88%"**를 썼다.
> v2는 판정에 쓰는 수치를 **전부 여기서 계산**한다.

````python
p = m["price"]
vc = p.value_counts()

# ── ④ 이산점(군집) ───────────────────────────────────────────────
print("[이산점]")
print(f"  고유값          : {p.nunique():,}개 / {len(p):,}행")
print(f"  상위 10개 값    : 전체의 {vc.head(10).sum() / len(p) * 100:.2f}%")
print(f"  상위 20개 값    : 전체의 {vc.head(20).sum() / len(p) * 100:.2f}%")

# 값이 소수 몇 자리까지 떨어지는가.
#  ⚠️ (p * 100 % 1 == 0) 으로 세면 부동소수점 오차 때문에 틀린 값이 나온다.
#     round(2) 와 원값을 비교하는 것이 안전하다.
print(f"  소수 2자리 이내 : {(p.round(2) == p).mean() * 100:.2f}%")
print()
print("  가장 흔한 가격 5개")
for v, n in vc.head(5).items():
    print(f"    ${v:>7.2f} : {n:>6,}명")

# ── 꼬리의 길이 ─────────────────────────────────────────────────
print()
print("[꼬리] 상위 분위수 — numerical_summary 는 75%까지만 준다")
for q in [0.90, 0.95, 0.99]:
    print(f"  {q:.0%} : ${p.quantile(q):>7.2f}")
print(f"  max : ${p.max():>7.2f}")
````

### 4-3-1. 봉우리 — 원척도와 log 척도를 나란히

**원척도로는 형태를 읽을 수 없다는 사실 자체를 수치로 남긴다.** 그래야 log 척도를 쓴 이유가 문서에 남는다.
두 히스토그램의 **구간 수를 12로 맞춰** 비교한다 — 구간 수가 다르면 정점 위치가 달라 보인다.

````python
BINS = 12

# ── 원척도 ──────────────────────────────────────────────────────
c0, e0 = histogram(p, bins=BINS)
print(f"[원척도 {BINS}구간] — 첫 구간에 얼마나 몰려 있나")
print(f"  ${e0[0]:.2f}~${e0[1]:.2f} : {c0[0]:,}명 ({c0[0] / len(p) * 100:.2f}%)")
print(f"  → 한 구간이 전체의 {c0[0] / len(p) * 100:.0f}%를 먹어 형태를 읽을 수 없다")

# ── log1p 척도 ──────────────────────────────────────────────────
print()
print(f"[log1p 척도 {BINS}구간] — 괄호 안은 원래 가격")
c, e = histogram(log1p(p), bins=BINS)
for i in range(BINS):
    bar = "█" * int(c[i] / max(c) * 40)
    print(f"  ${expm1(e[i]):6.2f}~${expm1(e[i+1]):7.2f} : {c[i]:>6,} {bar}")

# ── 봉우리 판정의 근거 ──────────────────────────────────────────
k = int(argmax(c))
up   = all(c[i] <= c[i+1] for i in range(0, k))        # 정점 앞이 계속 오르는가
down = all(c[i] >= c[i+1] for i in range(k, BINS - 1))  # 정점 뒤가 계속 내리는가

print()
print(f"  정점 구간      : ${expm1(e[k]):.2f}~${expm1(e[k+1]):.2f}  ({c[k]:,}명)")
print(f"  정점 앞 단조증가 : {up}")
print(f"  정점 뒤 단조감소 : {down}")
print(f"  → 둘 다 True 이면 이 구간 설정에서는 봉우리가 하나다(단봉)")
````

### 4-4. 로그변환이 실제로 효과가 있는가 〈추가된 셀〉

설계 §3-4가 요구한 확인이다 — **"변환 전후 왜도를 비교해 확인한다. 개선되지 않으면 변환하지 않는다."**

> **이것은 판정의 근거일 뿐 실행이 아니다.** 마스터 테이블의 `price` 컬럼은 건드리지 않는다.
> 실제 변환은 **⑤ 모델링용 전처리**에서, **분할 후**에 한다.

````python
print(f"변환 전 왜도      : {p.skew():>7.3f}")
print(f"log1p 적용 시     : {log1p(p).skew():>7.3f}")
print(f"log   적용 시     : {log(p).skew():>7.3f}")
print()
print(f"최솟값            : ${p.min()}  → 0을 포함하지 않으므로 log 를 쓸 수 있다")
print(f"helpers 권고      : {B_SUM.loc['price', 'log_need']}   (log_need 컬럼)")
print()
print("※ 우리 결정 : log 가 아니라 log1p 를 쓴다 (미결 ①, #10 검수 ①).")
print("   log1p 로도 기준 1.0 을 통과하고(0.805), 조회 변수 3종은 0 때문에 log1p 가 강제되므로")
print("   네 변수를 한 규칙으로 통일하는 편이 설명이 간단하다.")
````

## #06. ⚠️ `cart_weekday` — 요일 비율을 **그대로 읽으면 안 된다**

대상 기간은 **2019-10-11(금) ~ 11-24(일), 45일**이다.

```
45일 = 6주 + 3일
남는 사흘이 금·토·일  →  금·토·일은 7번 오고, 월~목은 6번 온다
```

그래서 **금요일 비율이 높아 보이는 것은 행동이 아니라 달력 때문**이다.

> 이 보정은 **단변량 분포 서술에만** 적용한다.
> Y 비율(구매율)은 **요일 내부의 비율**이라 관측 일수의 영향을 받지 않으므로 보정하지 않는다.

````python
WD = ["월", "화", "수", "목", "금", "토", "일"]

nday = (m["t0"].dt.normalize()          # 시각을 버리고 날짜만
                .drop_duplicates()      # 같은 날짜 한 번만
                .dt.dayofweek           # 0=월 … 6=일
                .value_counts()
                .sort_index()
                .to_list())

w = DataFrame({"고객 수":   [int((m["cart_weekday"] == x).sum()) for x in WD],
               "관측 일수": nday}, index=WD)

w["비율%"]  = (w["고객 수"] / len(m) * 100).round(2)
w["일평균"] = (w["고객 수"] / w["관측 일수"]).round(0).astype(int)

print("관측 일수 합계 :", sum(nday), "일   (45일이어야 한다)")
w
````

# #10. 검수 안내 — **배지환이 짚을 곳**

판정 문장은 Claude 초안이다. 아래를 하나씩 확인하고 동의하거나 고친다.
**수치가 아니라 "그래서 어떻게 할 것인가"를 본다.**

### ① `price` — `log`인가 `log1p`인가 → **결정됨 (임시, 이슬기 확인 대기)**

> **네 개 연속형 변수 전부 `log1p`로 통일한다.**

A조 결과가 나와 함께 정했다. A조 세 변수는 0이 **44.68~61.30%**라 `log`를 쓸 수 없고(`log(0)`이 정의되지 않는다),
`price`는 최솟값 \$0.05라 둘 다 가능하다. 그러나 **`log1p`로도 기준 1.0을 통과**하므로(0.805)
변환을 갈라 쓸 실익이 없고, 갈라 쓰면 *"왜 이건 log인가"*의 선택 기준이 **또 하나의 「교재 미기재」**가 된다.
교재는 로그변환 자체를 다루지 않으므로(`[LAB-07 §4-3]`) **변수별 최적화를 지지할 근거가 없다.**

**대가** — `price`의 변환 후 왜도가 0.055 대신 0.805가 된다. 다만 로지스틱 회귀의 가정 목록은
**로짓 선형성·VIF·독립성**이고(브리핑 §7) **X의 대칭성은 들어 있지 않으므로** 실질 손실이 아니다.

### ①-2 A조 미결 ① — 함께 결정된 것

`prior_view_cnt`(log1p 후 1.042)·`prior_uniq_product_cnt`(1.033)는 **변환 후에도 기준 1.0을 넘지만
그대로 변환한다.** 사유 넷 —
(1) 설계 §3-4의 두 조건(대상 `|왜도| ≥ 1.0` · 취소 `개선되지 않으면 안 함`)을 모두 통과했고,
**"변환 후 기준"은 설계에 없던 조항**이라 지금 추가하면 §12 위반이다.
(2) 잔차 왜도는 0 과다가 아니라 **실제 긴 꼬리**(최대 71회)에서 온다 — 0을 전부 빼면 오히려 1.643으로 악화한다.
(3) 로지스틱은 X의 분포를 가정하지 않는다.
(4) 기준 1.0은 상관계수 선택용으로 빌려온 관례선이고, 교재가 *"칼로 자르듯 판단 금지"*라 경고한 종류다 `[LAB-07 7단원 p.9]`.

**한계로 기록** — 두 변수는 변환 후에도 왜도가 1을 넘는다. 로짓 선형성은 ⑤ Box-Tidwell로 별도 확인한다.

### ② `price` — "이산성"을 어디까지 기록할 것인가

초안은 *"연속형으로 유지하되 완전한 연속은 아니라고 기록"*으로 두었다.
상위 20개 값이 28.81%를 덮는 것이 **로짓 선형성에 영향을 줄 수 있는지**는 ⑤의 Box-Tidwell이 답할 문제다.
지금 더 파고들지 않는 것이 설계 §3-10과 일관된다. **동의하는지 확인.**

### ③ `brand_grp` — `미상`을 별도 범주로 둘 것인가

초안은 **그대로 둔다**(06번 결정 유지). 다만 38.85%가 결측 라벨이라는 것은 작은 문제가 아니다.
**뒤집을 여지** — `미상`을 참조범주(reference category)로 지정하면 해석이 깔끔해질 수 있다.
이건 **⑤ 더미 인코딩 단계의 선택**이므로 여기서 정하지 않아도 되지만, **기록은 해 둔다.**

### ④ `cart_timeband` — 순서 정보를 버리는 것

초안은 더미 투입(순서 무시)으로 두었다. 교재가 *"서열척도를 적용하지 않더라도 모델링에 영향을 주지 않음"*
이라고 했으므로 근거는 있다 `[LAB-08 §3-6]`. **동의하는지 확인.**

### ⑤ A조와 합치기 전에

`09a_단변량_A조.csv`를 받아 이어 붙이면 **9개 변수 한 표**가 된다 `[LAB-08 2단원 p.22]` —
*"표로 분석 결과를 요약해서 모델링 설계의 근거로 삼는다."*
이 표가 **보고서 Ⅱ-1**과 **⑤ 로그변환 대상 목록**이 된다.

### ⑥ 검수 주체

설계 §4-2에 따라 **B조 결과는 이슬기가 검수한다** — `brand_grp`도 배지환이 06번에서 만들었기 때문이다.
